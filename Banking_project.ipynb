# Numerical analysis

```python
numerical_cols = ['Age', 'Estimated Income', 'Superannuation Savings',
                'Credit Card Balance', 'Bank Loans', 'Bank Deposits', 'Bank Deposits',
                'Checking Accounts', 'Saving Accounts', 'Business Lending']

# Univariate analysis
plot.figure(figsize=(15,15))
for i,col in enumerate(numerical_cols):
    plot.subplot(4,3,i+1)
    sns.histplot(df[col], kde=True)
    plot.title(col)

plot.tight_layout()
plot.show()
```

# Heatmap

```python
correlation_matrix = df[numerical_cols].corr()

plot.figure(figsize = (12,12))
sns.heatmap(correlation_matrix, annot = True, cmap = 'crest', fmt = ".2f")
plot.title('Correlation matrix')

# plot.savefig('EDA', dpi = 200)
plot.show()
```

# Insights on EDA

Correlation Matrix EDA Insights
1. Highly Correlated Features
	•	Bank Deposits, Checking Accounts, and Saving Accounts:
	•	Bank Deposits are highly correlated with Checking Accounts (0.84) and Saving Accounts (0.75).
	•	Two columns labeled “Bank Deposits” appear; both are perfectly correlated with each other (1.00), suggesting a possible duplicate or data redundancy.
	•	Business Lending:
	•	Moderately correlated with Bank Loans (0.42), Bank Deposits (0.44), Checking Accounts (0.36), and Saving Accounts (0.31).
2. Moderate Correlations
	•	Estimated Income:
	•	Moderately positively correlated with Superannuation Savings (0.37), Bank Loans (0.33), Bank Deposits (0.26-0.29), Checking Accounts (0.29), and Business Lending (0.33).
	•	Credit Card Balance:
	•	Shows moderate correlation with Bank Loans (0.37), Bank Deposits (0.38), Checking Accounts (0.30), and Business Lending (0.35).
3. Low or No Correlations
	•	Age:
	•	Age has very weak to negligible correlation with all other variables, mainly showing near 0 values.
	•	Superannuation Savings:
	•	Only moderate correlation with Estimated Income (0.37) and minor with others.
General Observations
	•	Banking Products Linkage: Customers with more deposits are likely to also have active checking and saving accounts, indicating product bundling or strong cross-ownership.
	•	Loan and Credit Products: Bank Loans and Credit Card Balances show moderate relationships with several deposit and income-related features, possibly reflecting financial behavior.
	•	Redundant Features: The presence of duplicate “Bank Deposits” columns should be examined—this may affect model or business analysis.
